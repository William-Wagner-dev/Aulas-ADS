PBI-04 - ESTRUTURA E MENU

```python
# Lista global para armazenar os pacientes.
# Cada paciente será um dicionário.
pacientes = []

def exibir_menu():
    """Exibe o menu de opções."""
    print("\n=== SISTEMA CLÍNICA VIDA+ ===")
    print("1. Cadastrar paciente")
    print("2. Ver estatísticas")
    print("3. Buscar paciente")
    print("4. Listar todos os pacientes")
    print("5. Verificar Acesso") #Lógica Booleana nova opção
    print("6. Sair")

def main():
    """Função principal do sistema."""
    while True:
        exibir_menu()

        # Bloco try-except para tratar erros de entrada que não sejam números
        try:
            opcao = input("Escolha uma opção: ")

            if opcao == '1':
                cadastrar_paciente() # A ser implementada
            elif opcao == '2':
                ver_estatisticas() # A ser implementada
            elif opcao == '3':
                buscar_paciente() # A ser implementada
            elif opcao == '4':
                listar_pacientes() # A ser implementada
            elif opcao == '5':
                verificar_acesso_paciente() #Chamada booleana
            elif opcao == '6':
                print("Encerrando o sistema. Até logo!")
                break # Sai do loop
            else:
                print("Opção inválida. Tente novamente.")

        except ValueError:
            print("Entrada inválida. Digite apenas o número da opção desejada.")

# if __name__ == "__main__":
#     main()
```

PBI-05 CADASTRO DE PACIENTES

```python
def cadastrar_paciente():
    """Permite cadastrar informações de um novo paciente."""
    print("\n--- Cadastro de Paciente ---")

    nome = input("Nome do paciente: ").strip()

    # Validação de Idade (Tratamento de Erro)
    while True:
        try:
            idade_input = input("Idade: ")
            idade = int(idade_input)
            if idade < 0 or idade > 150:
                 print("Idade inválida. Por favor, insira uma idade realista.")
                 continue
            break # Sai do loop se a idade for válida
        except ValueError:
            print(f"Erro: A idade deve ser um número. '{idade_input}' não é um número válido.")

    telefone = input("Telefone: ")

    novo_paciente = {
        'nome': nome,
        'idade': idade,
        'telefone': telefone
    }

    pacientes.append(novo_paciente)
    print("Paciente cadastrado com sucesso!")
```

PBI-06 ESTATÍSTICAS DA CLINICA

```python
def ver_estatisticas():
    """Calcula e exibe as estatísticas básicas dos pacientes."""

    if not pacientes:
        print("\nNenhum paciente cadastrado para calcular estatísticas.")
        return

    # 1. Número total de pacientes cadastrados
    total_pacientes = len(pacientes)
    print(f"\n--- Estatísticas da Clínica Vida+ ---")
    print(f"Número total de pacientes cadastrados: {total_pacientes} ")

    # Extrai as idades
    idades = [p['idade'] for p in pacientes]

    # 2. Idade média dos pacientes
    idade_media = sum(idades) / total_pacientes
    print(f"Idade média dos pacientes: {idade_media:.2f} anos [cite: 70]")

    # 3. Paciente mais novo e mais velho
    paciente_mais_novo = min(pacientes, key=lambda p: p['idade'])
    paciente_mais_velho = max(pacientes, key=lambda p: p['idade'])

    print(f"Paciente mais novo: {paciente_mais_novo['nome']} ({paciente_mais_novo['idade']} anos) ")
    print(f"Paciente mais velho: {paciente_mais_velho['nome']} ({paciente_mais_velho['idade']} anos) ")
```

PBI-07 BUSCA E LISTAGEM PACIENTES

```python
def buscar_paciente():
    """Permite buscar um paciente pelo nome."""
    if not pacientes:
        print("\nNenhum paciente cadastrado para buscar.")
        return

    nome_busca = input("\nDigite o nome (ou parte do nome) do paciente que deseja buscar: ").strip().lower()
    encontrados = [p for p in pacientes if nome_busca in p['nome'].lower()]

    if encontrados:
        print(f"\n--- {len(encontrados)} Paciente(s) Encontrado(s) ---")
        for i, p in enumerate(encontrados, 1):
            print(f"{i}. Nome: {p['nome']} | Idade: {p['idade']} | Tel: {p['telefone']}")
    else:
        print(f"Nenhum paciente encontrado com '{nome_busca}'.")

def listar_pacientes():
    """Exibe todos os pacientes cadastrados de forma organizada."""
    if not pacientes:
        print("\nNenhum paciente cadastrado.")
        return

    print(f"\n--- Lista de Todos os Pacientes ({len(pacientes)} no total) ---")
    print("{:<5} {:<25} {:<10} {:<15}".format("ID", "NOME", "IDADE", "TELEFONE"))
    print("-" * 55)

    for i, p in enumerate(pacientes, 1):
        print("{:<5} {:<25} {:<10} {:<15}".format(i, p['nome'], p['idade'], p['telefone']))
```

EXECUTANDO O SISTEMA

PBI-08 LOGICA BOOLEANA

```python
# Converte as expressões lógicas formais diretamente para a sintaxe Python.
# Usa 'and' para conjunção (^) e 'or' para disjunção (v).
def verificar_atendimento(A, B, C, D):
    """
    Verifica se o paciente pode ser atendido em ambas as modalidades
    baseado nas regras de lógica booleana.

    Args:
        A (bool): Agendamento marcado
        B (bool): Documentos em dia
        C (bool): Médico disponível
        D (bool): Pagamentos em dia

    Returns:
        tuple: (Pode_ter_Consulta_Normal, Pode_ter_Emergencia)
    """

    # Expressão Lógica para Consulta Normal (CN): (A ^ B ^ C) v (B ^ C ^ D)
    consulta_normal = (A and B and C) or (B and C and D)

    # Expressão Lógica para Emergência (E): C ^ (B v D)
    emergencia = C and (B or D)

    return consulta_normal, emergencia
```

PBI-08 - Coleta de dados

```python
# Interage com o usuário para converter entradas de texto ('V' ou 'F') em valores booleanos (True ou False).
def ler_status_booleano(pergunta):
    """Lê a entrada do usuário e retorna True/False."""
    while True:
        resposta = input(f"{pergunta} (S/N): ").strip().upper()
        if resposta == 'V':
            return True
        elif resposta == 'F':
            return False
        else:
            print("Entrada inválida. Digite apenas 'V' para Verdadeiro ou 'F' para Falso.")

def verificar_acesso_paciente():
    """Coleta as variáveis lógicas e executa a verificação de atendimento."""
    print("\n--- Verificação de Controle de Acesso ---")

    # Coleta das variáveis
    A = ler_status_booleano("A: Paciente tem agendamento marcado?")
    B = ler_status_booleano("B: Paciente está com documentos em dia?")
    C = ler_status_booleano("C: Há médico disponível no horário?")
    D = ler_status_booleano("D: Paciente está em dia com pagamentos anteriores?")

    # Executa a lógica booleana
    pode_normal, pode_emergencia = verificar_atendimento(A, B, C, D)

    print("\n--- RESULTADO DO ATENDIMENTO ---")

    # Exibe o resultado
    if pode_normal:
        print("✅ Atendimento liberado como: CONSULTA NORMAL.")
    else:
        print("❌ Atendimento negado como: CONSULTA NORMAL.")

    if pode_emergencia:
        print("✅ Atendimento liberado como: EMERGÊNCIA.")
    else:
        print("❌ Atendimento negado como: EMERGÊNCIA.")

    # Análise extra (Caso o paciente não tenha sido liberado em nenhuma modalidade)
    if not pode_normal and not pode_emergencia:
        print("\n🚫 ALERTA: Paciente não pode ser atendido em nenhuma modalidade.")
```

"PSEUDOCODIGO" de Fila de espera

```python
#Tomei a liberdade de pesquisar, estudar e desenvolver o "pseudocódigo"
#Na linguagem PYTHON

def simular_fila_atendimento():
    """
    Implementa o algoritmo de Fila (Queue) FIFO em Python.
    Permite inserir 3 pacientes, remover o primeiro para atendimento e listar os remanescentes.
    """

    # Abordagem: A lista 'fila_atendimento' será usada como a nossa Fila FIFO.
    fila_atendimento = []

    print("\n--- PASSO 4: SIMULAÇÃO DE FILA DE ATENDIMENTO (FIFO) ---")

    # 1. INSERIR 3 PACIENTES NA FILA (ENQUEUE)

    # Abordagem: O loop 'range(3)' garante que a inserção seja feita para 3 pacientes.
    for i in range(1, 4):
        print(f"\n--- Inserção de Paciente {i} ---")

        # Coleta de Dados
        nome = input("Informe o Nome do paciente: ").strip()
        cpf = input("Informe o CPF do paciente: ").strip()

        # Criação do Dicionário (Registro)
        novo_paciente = {
            'nome': nome,
            'cpf': cpf
        }

        # Adiciona ao final da fila (ENQUEUE: list.append())
        fila_atendimento.append(novo_paciente)
        print(f"Paciente '{nome}' adicionado ao final da fila.")

    print("\n--- FILA INICIAL COMPLETA ---")
    print(f"Pacientes na Fila: {len(fila_atendimento)}")

    # 2. REMOVER O PRIMEIRO PACIENTE PARA ATENDIMENTO (DEQUEUE)

    if fila_atendimento:
        # Abordagem: O método pop(0) remove e retorna o primeiro elemento,
        # garantindo que o "primeiro a entrar" seja o "primeiro a sair" (FIFO).
        paciente_chamado = fila_atendimento.pop(0)

        print("\n--- ATENDIMENTO LIBERADO (DEQUEUE) ---")
        print(f"Paciente Chamado: {paciente_chamado['nome']} (CPF: {paciente_chamado['cpf']})")
        print("O paciente saiu da fila.")
    else:
        print("\nA fila de atendimento está vazia.")

    # 3. MOSTRAR QUEM AINDA ESTÁ NA FILA

    print("\n--- STATUS ATUAL DA FILA DE ESPERA ---")

    if fila_atendimento:
        print(f"Número de pacientes restantes na fila: {len(fila_atendimento)}")

        # Abordagem: Loop simples para iterar sobre os elementos remanescentes da lista (Fila)
        for i, paciente in enumerate(fila_atendimento, 1):
            print(f"  {i}. Nome: {paciente['nome']} (CPF: {paciente['cpf']})")
    else:
        print("A fila de atendimento está vazia após o atendimento.")

# Chamada da função para execução do Passo 4
if __name__ == "__main__":
    simular_fila_atendimento()
```
